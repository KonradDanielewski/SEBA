```python
import units_processing as up
import behaviour_annotations as ba
import stimulus_annotations as sa

from pynwb import NWBHDF5IO

# Create the NWB file from the .bvs filename
nwb = up.create_nwbfile_from_bvs_path(behaviour_path)

# Populate Units from Kilosort/Phy outputs + cluster_info.csv
up.populate_nwb_units_from_kilosort_files(
    nwb=nwb,
    spike_times_path=spike_times_path,
    spike_clusters_path=spike_clusters_path,
    sampling_rate=npx_sampling_rate,
    cluster_info_path=cluster_info_path,
    UNIT_COL_DESCRIPTIONS=UNIT_COL_DESCRIPTIONS,
)

nwb = ba.add_bvs_behavior_annotations_from_paths(
    nwb,
    behaviour_path=behaviour_path,
    cam_TTL_path=cam_TTL_path,
    fps_video=video_fps,
)
nwb = sa.add_cs_onsets_from_path(nwb, cs_path)

# ---------- Write the NWB file ----------
out_path = "session_min.nwb"
with NWBHDF5IO(out_path, "w") as io:
    io.write(nwb)

print(f"Wrote {out_path}")

nwb
```
